```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels as sm
```

### Creating a patient Record File To Analyse

```python
data = pd.read_csv("Patient.csv",sep="\t")
display(data)

data.info()
data.describe()
```

### Most Health condition in the Hospital

```python
a = data["HEALTH CONDITION"].value_counts()
display(pd.DataFrame(a))
data["HEALTH CONDITION"].value_counts().plot(kind="bar")
plt.show()
```

### Creating the data  on the drug product in the hospital

```python
b=pd.DataFrame(data["DRUG PRODUCT"].value_counts())
display(b)
# Creating the visual display of the data
data["DRUG PRODUCT"].value_counts().plot(kind="bar")
plt.show()
```

### Creating a data of which drug is used for which conditions

```python
q =pd.DataFrame(data.groupby("HEALTH CONDITION")["DRUG PRODUCT"].value_counts())
display(q)

q.plot(kind="bar")
plt.savefig("HEALTH CONDITION.png")
plt.show()
```
